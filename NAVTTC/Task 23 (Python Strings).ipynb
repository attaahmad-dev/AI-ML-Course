## Python Strings

#### create string type variables

```python
name = "Python"
print(name)

message = "I love Python."
print(message)
```

#### Access String Character in Python

```python
greet = 'hello'

# access 1st index element
print(greet[1]) # "e"
```

#### Negative Indexing

```python
greet = 'hello'

# access 4th last element
print(greet[-4]) # "e"
```

#### Slicing

```python
greet = 'Hello'

# access character from 1st index to 3rd index
print(greet[1:4])  # "ell"
```

#### Python Strings are Immutable

```python
message = 'Hola Amigos'
message[0] = 'H'
print(message)
```

#### However, we can assign the variable name to a new string
For example

```python
message = 'Hola Amigos'

# assign new string to message variable
message = 'Hello Friends'

print(message); # prints "Hello Friends"
```

#### Python Multiline Strings
Multiline string

```python
message = """
Never gonna give you up
Never gonna let you down
"""

print(message)
```

#### Python String Operation

```python
# 1. Compare Two Strings
str1 = "Hello, world!"
str2 = "I love Swift."
str3 = "Hello, world!"

# compare str1 and str2
print(str1 == str2)

# compare str1 and str3
print(str1 == str3)
```

```python
#Join two or more Strings
greet = "Hello, "
name = "Jack"

# using + operator
result = greet + name
print(result)

# Output: Hello, Jack
```

#### Iterate through a Python String

```python
greet = 'Hello'

# iterating through greet string
for letter in greet:
    print(letter)
```

#### Python String Length

```python
greet = 'Hello'

# count length of greet string
print(len(greet))

# Output: 5
```

#### String Membership Test

```python
print('a' in 'program') # True
print('at' not in 'battle') # False
```

#### Escape Sequences in Python

```python
example = "He said, "What's there?""

print(example) # throws error
```

#### Solving the issue
escape double quotes

```python
example = "He said, \"What's there?\""

# escape single quotes
example = 'He said, "What\'s there?"'

print(example)

# Output: He said, "What's there?"
```

#### Python String Formatting (f-Strings)

```python
name = 'Cathy'
country = 'UK'

print(f'{name} is from {country}')
```

**Challenge:
Write a function to double every letter in a string.**

For input 'hello', the return value should be 'hheelllloo'.

```python
def double_letters(s):
    result = ""
    for char in s:
        result += char * 2
    return result

print(double_letters("hello"))
```
